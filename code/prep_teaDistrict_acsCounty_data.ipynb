```python
dataset_path = '../feature_inputs/'

enroll_acs_regr = pd.read_csv(dataset_path+"acs_enrollment_percent_regression.csv")
enroll_edu_acs_regr = pd.read_csv(dataset_path+"acs_enroll_education_percent_regression.csv")

enroll_tea_regr = pd.read_csv(dataset_path+"tea_districts_enroll_percent_regression.csv")
enroll_sal_tea_regr = pd.read_csv(dataset_path+"tea_districts_enroll_percent_staff_salary_regression.csv")
```

```python
#enroll_acs_regr = enroll_acs_regr.add_suffix('_ACS')
#enroll_edu_acs_regr = enroll_edu_acs_regr.add_suffix('_ACS')
#enroll_tea_regr = enroll_tea_regr.add_suffix('_TEA')
#enroll_sal_tea_regr = enroll_sal_tea_regr.add_suffix('_TEA')

acs_drop_columns = ['County', 'MMR Vaccination Rate']
tea_drop_columns_corr = ['County_Name', 'District_Name', 'Total_Count', 'MMR Vaccination Rate']

enroll_acs_regr = enroll_acs_regr.rename(columns= {col: f"{col}_ACS" for col in enroll_acs_regr.drop(columns=acs_drop_columns).columns})
enroll_edu_acs_regr = enroll_edu_acs_regr.rename(columns={col: f"{col}_ACS" for col in enroll_edu_acs_regr.drop(columns=acs_drop_columns).columns})
enroll_tea_regr = enroll_tea_regr.rename(columns={col: f"{col}_TEA" for col in enroll_tea_regr.drop(columns=tea_drop_columns_corr).columns})
enroll_sal_tea_regr = enroll_sal_tea_regr.rename(columns={col: f"{col}_TEA" for col in enroll_sal_tea_regr.drop(columns=tea_drop_columns_corr).columns})
```
