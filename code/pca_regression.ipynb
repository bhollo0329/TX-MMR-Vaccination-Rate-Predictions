```python
n = 10
scaler = MinMaxScaler()
pca_model = PCA(n_components=n)
y_data = all_vac_set_no_gender_percent['MMR Vaccination Rate']
x_data = all_vac_set_no_gender_percent.drop(columns=['MMR Vaccination Rate', 'County'])
x_scaler = scaler.fit_transform(x_data)
x_pca = pca_model.fit_transform(x_scaler)
#modelR = Ridge(alpha=1.0).fit(x_pca, y_data)
```

```python
coef_list= []
for i in range(10):
    x_boot, y_boot = resample(x_pca, y_data)
    modelR = Ridge(alpha=1.0).fit(x_boot, y_boot)
    coef_list.append(modelR.coef_)
```

```python
component_names = ['PC'+str(x) for x in range(n)]
df = pd.DataFrame(coef_list,  columns=component_names)
```

```python
ci_lower = np.percentile(df, 2.5, axis=0)
ci_upper = np.percentile(df, 97.5, axis=0)
coef_mean = np.mean(df, axis=0).to_numpy()

ci_results_df = pd.DataFrame({'Predictor': component_names,
                                            'Mean': coef_mean,
                                            'CI_Lower': ci_lower,
                                            'CI_Upper': ci_upper})
    
ci_results_df["Explained Variance"] = pca_model.explained_variance_ratio_
ci_results_df['Coeff Signficance'] = ['Not Signficant' if l <= 0 <= u else 'Signficant' for l,u in zip(ci_results_df['CI_Lower'], ci_results_df['CI_Upper'])]
```
